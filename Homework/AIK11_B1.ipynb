### **Bài tập 1.1: Trợ lý AI tính chi phí huấn luyện mô hình**

*   **Bối cảnh:** Bạn đang viết một chương trình nhỏ cho trợ lý AI để giúp một công ty khởi nghiệp tính toán ngân sách khi thuê máy chủ đám mây (Cloud Server) để huấn luyện mô hình AI.
*   **Yêu cầu:**
    1. Dùng lệnh `input()` để hỏi người dùng nhập vào **Số lượng GPU cần thuê** (ép kiểu sang số nguyên `int`).
    2. Nhập vào **Đơn giá thuê mỗi GPU/giờ** (ép kiểu sang số thực `float`).
    3. Nhập vào **Số giờ dự kiến huấn luyện** (ép kiểu sang số thực `float`).
    4. Nhập vào **Số tiền ngân sách tối đa** mà công ty đang có (ép kiểu sang số thực `float`).
*   **Các phép tính cần làm:**
    *   **Phép nhân (`*`):** Tính **Tổng chi phí dự kiến** = `Số GPU` \\(\times\\) `Đơn giá` \\(\times\\) `Số giờ`.
    *   **Phép trừ (`-`):** Tính **Số tiền còn lại** trong ngân sách sau khi trừ đi chi phí = `Ngân sách tối đa` \\(-\\) `Tổng chi phí dự kiến`.
    *   **Phép chia (`/`):** Nếu công ty có **3 phòng ban** cùng chia đều chi phí này, hãy tính **Số tiền mỗi phòng ban phải trả** = `Tổng chi phí dự kiến` \\(/\\) `3`.
*   **Đầu ra mong muốn:** Sử dụng `print()` để xuất ra màn hình báo cáo chi tiết của trợ lý AI (ví dụ: in ra tổng chi phí, số tiền còn lại trong ngân sách và số tiền mỗi phòng ban cần đóng góp).



Nhập dữ liệu

```python
print("Chào mừng bạn đến với Trợ lý AI tính chi phí huấn luyện mô hình!")
so_luong_gpu = int(input("Nhập số lượng GPU cần thuê: "))
don_gia = float(input("Nhập đơn giá thuê mỗi GPU/giờ: "))
so_gio = float(input("Nhập số giờ dự kiến huấn luyện: "))
ngan_sach_toi_da = float(input("Nhập số tiền ngân sách tối đa mà công ty đang có: "))
```

Tính toán chi phí

```python
# tính chi phí
tong_chi_phi = so_luong_gpu * don_gia * so_gio # tổng chi phí sử dụng
tien_con_lai = ngan_sach_toi_da - tong_chi_phi
tien_moi_phong = tong_chi_phi / 3
```

```python
print("\n--- BÁO CÁO CHI TIẾT TỪ TRỢ LÝ AI ---")
print(f"Tổng chi phí dự kiến: {tong_chi_phi:,.2f} VNĐ")
print(f"Số tiền còn lại trong ngân sách: {tien_con_lai:,.2f} VNĐ")
print(f"Số tiền mỗi phòng ban phải trả (chia đều cho 3 phòng): {tien_moi_phong:,.2f} VNĐ")
```

### **Bài tập 1.2: AI đánh giá hiệu suất Chatbot**

*   **Bối cảnh:** Bạn lập trình cho một hệ thống đánh giá chatbot tự động chăm sóc khách hàng sau một ngày hoạt động.
*   **Yêu cầu:**
    1. Dùng lệnh `input()` yêu cầu người dùng nhập **Tổng số câu hỏi** mà khách hàng đã gửi trong ngày (ép kiểu số nguyên `int`).
    2. Yêu cầu nhập **Số câu trả lời chưa chính xác** của chatbot (ép kiểu số nguyên `int`).
*   **Các phép tính cần làm:**
    *   **Phép trừ (`-`):** Tính **Số câu trả lời đúng** = `Tổng số câu hỏi` \\(-\\) `Số câu trả lời chưa chính xác`.
    *   **Phép chia (`/`) và nhân (`*`):** Tính **Tỷ lệ chính xác (Accuracy) theo %** = `(Số câu trả lời đúng / Tổng số câu hỏi) * 100`.
*   **Đầu ra mong muốn:** Sử dụng `print()` để in ra kết quả báo cáo của chatbot:
    `"Báo cáo hiệu suất: Chatbot đã trả lời đúng [Số câu đúng] câu hỏi. Đạt tỷ lệ chính xác [Tỷ lệ %]%."`
