### **Bài tập 2.2: AI phân loại Email rác - Spam Filter (Luyện tập so sánh số)**

*   **Bối cảnh:** Bạn lập trình cho một bộ lọc thư rác thông minh. AI phân tích email và đánh giá **điểm Spam Score** theo thang điểm từ 1 đến 10 (điểm càng cao khả năng là thư rác càng lớn).
*   **Yêu cầu:**
    1. Yêu cầu người dùng nhập vào **Điểm Spam Score** của email vừa nhận (ép kiểu số nguyên `int`).
    2. Sử dụng cấu trúc **`if-else`** để kiểm tra:
        *   Nếu điểm số **lớn hơn hoặc bằng 5**, in ra: `"Thư rác (Spam) - Đã tự động chuyển vào thùng rác."`
        *   Nếu điểm số **nhỏ hơn 5**, in ra: `"Thư thường (Inbox) - Hiển thị trong hộp thư đến."`



### **Bài tập 2.3: Hệ thống giám sát nhiệt độ GPU huấn luyện AI (Luyện tập `if-elif-else`)**

*   **Bối cảnh:** Máy chủ huấn luyện AI hoạt động liên tục khiến chip xử lý đồ họa (GPU) bị nóng lên. Bạn cần viết một chương trình giám sát nhiệt độ tự động để bảo vệ phần cứng.
*   **Yêu cầu:**
    1. Yêu cầu người dùng nhập vào **Nhiệt độ hiện tại của GPU** tính theo độ C (ép kiểu số thực `float` hoặc số nguyên `int`).
    2. Sử dụng cấu trúc **`if-elif-else`** để phân loại các mức độ:
        *   Nếu nhiệt độ **dưới 65 độ C**, in ra: `"Trạng thái: GPU hoạt động mát mẻ."`
        *   Nếu nhiệt độ **từ 65 đến 80 độ C**, in ra: `"Trạng thái: GPU hơi nóng. Hệ thống tự động tăng tốc độ quạt gió."`
        *   Nếu nhiệt độ **trên 80 độ C**, in ra: `"CẢNH BÁO: GPU quá nhiệt! Tự động ngắt tiến trình huấn luyện AI để bảo vệ phần cứng!"`


### **Bài tập 3.2: AI kiểm tra sản phẩm lỗi trực tiếp (Sử dụng `for ... in range()`, `if` và biến đếm)**

*   **Bối cảnh:** Thay vì dùng danh sách có sẵn, camera AI tại nhà máy sẽ quét lần lượt **4 sản phẩm** chạy qua băng chuyền và yêu cầu bạn nhập điểm đánh giá trực tiếp cho từng sản phẩm một.
*   **Yêu cầu:**
    1. Tạo một biến đếm ban đầu: `so_san_pham_loi = 0`.
    2. Sử dụng vòng lặp `for i in range(1, 5)` để giả lập quét 4 sản phẩm.
    3. Ở mỗi lượt, dùng lệnh `input()` yêu cầu nhập vào **Điểm chất lượng** của sản phẩm đó (ép kiểu số nguyên `int` từ 1 đến 10).
    4. Sử dụng cấu trúc `if` để kiểm tra: Nếu điểm số **nhỏ hơn 5**, in ra màn hình: `"Sản phẩm lỗi!"` và **cộng thêm 1** vào biến `so_san_pham_loi`.
    5. Sau khi kết thúc vòng lặp, dùng lệnh `print()` để in ra tổng số sản phẩm lỗi mà AI phát hiện được.



### **Bài tập 3.3: Khóa bảo mật của trợ lý AI (Sử dụng `while`, biến đếm và `if ... break`)**

*   **Bối cảnh:** Trợ lý ảo AI yêu cầu người dùng nhập đúng mật mã để kích hoạt hệ thống. Để bảo mật, AI chỉ cho phép thử tối đa **3 lần**.
*   **Yêu cầu:**
    1. Khởi tạo một biến đếm: `so_lan_thu = 0`.
    2. Sử dụng vòng lặp `while so_lan_thu < 3`:
        *   Dùng `input()` yêu cầu người dùng nhập mật mã kích hoạt (kiểu chuỗi `str`).
        *   Cộng thêm 1 vào biến `so_lan_thu`.
        *   Sử dụng cấu trúc `if-else` để kiểm tra:
            *   Nếu mật mã nhập vào là `"AI-Open"`, in ra: `"Kích hoạt trợ lý AI thành công!"` và dùng lệnh **`break`** để thoát vòng lặp.
            *   Ngược lại (nhập sai), in ra: `"Mật mã không đúng!"`
    3. Sau khi kết thúc vòng lặp, nếu người dùng đã dùng hết cả 3 lần thử mà vẫn không đúng mật mã, hãy in ra cảnh báo: `"Tài khoản bị khóa tạm thời do nhập sai quá 3 lần!"`
