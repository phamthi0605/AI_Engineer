### **Bài tập 2: Tính tỷ lệ chính xác (Accuracy) của Mô hình AI (Hàm có `return`)**
*   **Bối cảnh:** Bạn cần một hàm tính toán nhanh tỷ lệ phần trăm chính xác của mô hình phân loại sau khi đánh giá.
*   **Yêu cầu:**
    1. Định nghĩa hàm `tinh_accuracy(so_cau_dung, tong_so_cau)`.
    2. Trong hàm, tính toán `accuracy = (so_cau_dung / tong_so_cau) * 100` và trả về kết quả bằng từ khóa **`return`**.
    3. Ở chương trình chính, gọi hàm với `so_cau_dung = 85` và `tong_so_cau = 100`, sau đó in ra kết quả trả về.



### **Bài tập 4: Lọc và đếm vật thể nhận diện hợp lệ (AI Object Detection)**

*   **Bối cảnh:** Một AI camera giao thông quét qua một khung hình và trả về danh sách điểm độ tin cậy (confidence scores từ `0.0` đến `1.0`) của các vật thể phát hiện được. Bạn cần viết một hàm đóng gói logic để tự động đếm xem có bao nhiêu vật thể đạt chuẩn độ tin cậy để hiển thị lên màn hình.

*   **Yêu cầu:**
    1. Định nghĩa hàm **`dem_vat_the_hop_le(scores, nguong_tin_cay=0.7)`** nhận vào 2 tham số:
        *   `scores`: danh sách các điểm độ tin cậy (kiểu `list`).
        *   `nguong_tin_cay`: ngưỡng điểm tối thiểu để công nhận vật thể (có giá trị mặc định là `0.7`).
    2. Bên trong hàm:
        *   Tạo một biến đếm ban đầu `so_luong = 0`.
        *   Sử dụng vòng lặp **`for`** để duyệt qua từng điểm số trong `scores`.
        *   Sử dụng câu lệnh **`if`** để kiểm tra: Nếu điểm số **lớn hơn hoặc bằng `nguong_tin_cay`**, hãy cộng thêm `1` vào biến `so_luong`.
        *   Sử dụng **`return`** để trả về tổng `so_luong` vật thể hợp lệ.
    3. Ở chương trình chính:
        *   Cho sẵn danh sách: `danh_sach_diem = [0.95, 0.42, 0.88, 0.65, 0.72, 0.30]`
        *   Gọi hàm với danh sách trên và in kết quả ra màn hình.
