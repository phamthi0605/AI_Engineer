### **Bài tập 1: AI lọc điểm độ tin cậy (Confidence Score Filter)**
*   **Bối cảnh:** Một mô hình AI nhận diện hình ảnh trả về danh sách các điểm độ tin cậy (thang điểm từ `0.0` đến `1.0`) cho từng vật thể phát hiện được trong khung hình.
*   **Dữ liệu ban đầu:** `scores = [0.92, 0.45, 0.88, 0.30, 0.75]`
*   **Yêu cầu:**
    1. Dùng vòng lặp `for` để duyệt qua từng điểm số trong danh sách `scores`.
    2. Dùng `if-else` để kiểm tra:
        *   Nếu điểm số **lớn hơn hoặc bằng `0.7`**: In ra `"Vật thể hợp lệ (Độ tin cậy: [điểm số])"`.
        *   Ngược lại: In ra `"Nhiễu/Bỏ qua (Độ tin cậy: [điểm số])"`.


```python
scores = [0.92, 0.45, 0.88, 0.30, 0.75]
for i in range(len(scores)):
    if scores[i] >= 0.7:
        print(f"Vật thể hợp lệ (Độ tin cậy: {scores[i]})")
    else:
         print(f"Nhiễu/ Bỏ qua (Độ tin cậy: {scores[i]})")
```


### **Bài tập 2: Đếm phản hồi tiêu cực cho Chatbot (Sentiment Analysis)**
*   **Bối cảnh:** Hệ thống AI phân tích cảm xúc khách hàng duyệt qua danh sách các nhãn đánh giá sau cuộc trò chuyện với Chatbot.
*   **Dữ liệu ban đầu:** `feedbacks = ["Positive", "Negative", "Positive", "Negative", "Negative"]`
*   **Yêu cầu:**
    1. Khởi tạo một biến đếm `so_luong_negative = 0`.
    2. Dùng vòng lặp `for` duyệt qua danh sách `feedbacks`.
    3. Dùng câu lệnh `if` kiểm tra: Nếu phản hồi bằng `"Negative"`, hãy tăng `so_luong_negative` lên `1`.
    4. Sau khi duyệt xong, in ra tổng số phản hồi tiêu cực: `"Tổng số đánh giá tiêu cực: [số lượng]"`.


```python
feedbacks = ["Positive", "Negative", "Positive","Negative", "Negative" ]
so_luong_negative = 0
for i in range(len(feedbacks)):
    if feedbacks[i] == "Negative":
        so_luong_negative += 1
print(f"Số feedback tiêu cực:[{so_luong_negative}]")
```


### **Bài tập 3: AI kiểm duyệt nội dung vi phạm (Content Moderator)**
*   **Bối cảnh:** Hệ thống AI kiểm duyệt văn bản tự động quét qua danh sách các từ trong một đoạn bình luận của người dùng để tìm từ cấm.
*   **Dữ liệu ban đầu:** `words = ["xin chào", "spam", "học tập", "cấm", "python"]`
*   **Yêu cầu:**
    1. Dùng vòng lặp `for` duyệt qua từng từ trong danh sách `words`.
    2. Dùng `if` để kiểm tra: Nếu từ đó là `"spam"` hoặc `"cấm"`, in ra cảnh báo: `"Cảnh báo: Phát hiện từ vi phạm '[từ]'!"`.


```python
words = ["xin chào", "spam", "học tập", "cấm", "python"]
for i in range(len(words)):
    if words[i] == "spam" or words[i] == "cấm":
        print(f"Cảnh báo: Phát hiện từ vi phạm '[{words[i]}]'!")
```


### **Bài tập 4: Giám sát tốc độ phản hồi của trợ lý AI (Response Time)**
*   **Bối cảnh:** Hệ thống đo thời gian phản hồi (tính bằng giây) của 5 lần gọi API liên tiếp tới trợ lý ảo.
*   **Dữ liệu ban đầu:** `response_times = [0.5, 2.3, 0.8, 3.1, 1.2]`
*   **Yêu cầu:**
    1. Duyệt qua từng khoảng thời gian trong danh sách `response_times`.
    2. Dùng `if-else` để phân loại:
        *   Nếu thời gian **lớn hơn `2.0` giây**: In ra `"Thời gian [thời gian]s - Phản hồi chậm!"`.
        *   Ngược lại: In ra `"Thời gian [thời gian]s - Phản hồi nhanh"`.


```python
response_time = [0.5, 2.3, 0.8, 3.1, 1.2]
for i in range(len(response_time)):
    if response_time[i] > 2:
        print(f"Thời gian [{response_time[i]}]s - Phản hồi chậm")
    else:
        print(f"Thời gian [{response_time[i]}]s - Phản hồi nhanh")
```
